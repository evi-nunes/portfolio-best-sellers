```python
import pandas as pd
import numpy as np

# Carrega os dados
df = pd.read_csv('../dados/summer-products-with-rating-and-performance_2020-08.csv')

# Mostra informações gerais
print(f"Total de linhas: {df.shape[0]}")
print(f"Total de colunas: {df.shape[1]}")
print("\nPrimeiras linhas:")
df.head()
```

```python
# Verifica quantos valores faltam em cada coluna
print("Valores faltantes por coluna:\n")
missing = df.isnull().sum()
print(missing)

print("\n\nPercentual de dados faltantes:")
percent_missing = (df.isnull().sum() / len(df)) * 100
print(percent_missing)
```

```python
# Preenche valores faltantes nas colunas de rating com 0
colunas_rating = ['rating_five_count', 'rating_four_count', 'rating_three_count', 
                  'rating_two_count', 'rating_one_count']

df[colunas_rating] = df[colunas_rating].fillna(0)

# Para product_color e product_variation_size_id, vamos deletar essas linhas
df = df.dropna(subset=['product_color', 'product_variation_size_id'])

print("Valores faltantes após limpeza:")
print(df.isnull().sum().sum())
print(f"\nTotal de linhas após limpeza: {df.shape[0]}")
```

```python
# Mostra quais colunas ainda têm valores faltantes
print("Valores faltantes restantes:\n")
missing_after = df.isnull().sum()
print(missing_after[missing_after > 0])
```

```python
# Deleta linhas com valores faltantes em colunas não essenciais
df = df.dropna(subset=['has_urgency_banner', 'urgency_text', 'merchant_profile_picture'])

print("Valores faltantes após deletar colunas não essenciais:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print(f"\nTotal de linhas agora: {df.shape[0]}")
```

```python
# Preenche origin_country faltantes com "Unknown"
df['origin_country'] = df['origin_country'].fillna('Unknown')

print("Valores faltantes finais:")
print(df.isnull().sum()[df.isnull().sum() > 0])
if df.isnull().sum().sum() == 0:
    print("\n✓ Dados completamente limpos! Sem valores faltantes.")
    
print(f"\nTotal final de linhas: {df.shape[0]}")
```

```python
# Verifica preços
print("Estatísticas de PREÇO:")
print(df['price'].describe())
print(f"\nPreço mínimo: {df['price'].min()}")
print(f"Preço máximo: {df['price'].max()}")

# Verifica vendas
print("\n\nEstatísticas de VENDAS:")
print(df['units_sold'].describe())
print(f"\nVendas mínimo: {df['units_sold'].min()}")
print(f"Vendas máximo: {df['units_sold'].max()}")
```

```python
# Detecta outliers usando IQR (Interquartile Range)
def detectar_outliers(coluna):
    Q1 = df[coluna].quantile(0.25)
    Q3 = df[coluna].quantile(0.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    
    outliers = df[(df[coluna] < limite_inferior) | (df[coluna] > limite_superior)]
    return len(outliers)

print("Outliers detectados:")
print(f"Preço: {detectar_outliers('price')} outliers")
print(f"Vendas: {detectar_outliers('units_sold')} outliers")
print(f"Rating: {detectar_outliers('rating')} outliers")
```

```python
# Deleta apenas outliers de preço
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

df = df[(df['price'] >= limite_inferior) & (df['price'] <= limite_superior)]

print(f"Total de linhas após remover outlier de preço: {df.shape[0]}")
print(f"\nDados completamente limpos e prontos para análise!")
```

```python
# Salva os dados limpos em um novo CSV
df.to_csv('../dados/summer-products-cleaned.csv', index=False)

print("✓ Dados limpos salvos em: dados/summer-products-cleaned.csv")
print(f"\nResumo da Limpeza:")
print(f"- Valores faltantes removidos")
print(f"- Outliers de preço removidos")
print(f"- Total final: {df.shape[0]} produtos prontos para análise")
```
